```python
# ----- DATA -----
x = 2.0
y = 10.0   # true value

# ----- WEIGHTS (Input → Hidden) -----
w11 = 0.5
w12 = -0.3

b1 = 0.1
b2 = 0.2

# ----- WEIGHTS (Hidden → Output) -----
w21 = 0.7
w22 = -1.2

b3 = 0.3

learning_rate = 0.01

# ----- TRAINING LOOP -----
for epoch in range(10):

    # ===== FORWARD PASS =====
    z1 = w11 * x + b1
    z2 = w12 * x + b2

    # ReLU
    a1 = max(0, z1)
    a2 = max(0, z2)

    # Output
    y_hat = w21 * a1 + w22 * a2 + b3

    # ===== LOSS =====
    loss = (y_hat - y) ** 2

    # ===== BACKWARD PASS =====

    # Output gradients
    dL_dyhat = 2 * (y_hat - y)

    dL_dw21 = dL_dyhat * a1
    dL_dw22 = dL_dyhat * a2
    dL_db3  = dL_dyhat

    # Hidden gradients
    dL_da1 = dL_dyhat * w21
    dL_da2 = dL_dyhat * w22

    # ReLU backward
    dL_dz1 = dL_da1 * (1 if z1 > 0 else 0)
    dL_dz2 = dL_da2 * (1 if z2 > 0 else 0)

    # Input layer gradients
    dL_dw11 = dL_dz1 * x
    dL_dw12 = dL_dz2 * x

    dL_db1 = dL_dz1
    dL_db2 = dL_dz2

    # ===== UPDATE =====
    w21 -= learning_rate * dL_dw21
    w22 -= learning_rate * dL_dw22
    b3  -= learning_rate * dL_db3

    w11 -= learning_rate * dL_dw11
    w12 -= learning_rate * dL_dw12
    b1  -= learning_rate * dL_db1
    b2  -= learning_rate * dL_db2

    # ===== PRINT =====
    print(f"Epoch {epoch}, Loss: {loss:.4f}, Prediction: {y_hat:.4f}")
```
